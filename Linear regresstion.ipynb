```python
#Simpal linear regresstion(student marks predication)
#Y=mX+c, 
#Y= output, X= Input, m=slop(how much Y changes when X increases by 1) c → Intercept (value of Y when X = 0)
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Data
X = np.array([[2], [4], [6], [8]])
Y = np.array([40, 55, 70, 85])

# Model
model = LinearRegression()
model.fit(X, Y)

# Prediction
print(model.predict([[5]]))
```

```python
#Multiple Linear Regression(House price prediction)
#Y=b0+b1X1+b2X2+b3X3
#Where:	Y → Predicted output
#b₀ → Intercept
#	b₁, b₂, b₃ → Coefficients (tells us how strongly an input affects the output)
#	X₁, X₂, X₃ → Independent variables
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# Create house dataset
data = {
    'Area': [800, 1000, 1200, 1500, 1800],
    'Bedrooms': [2, 2, 3, 3, 4],
    'Age': [20, 15, 10, 5, 2],
    'Price': [50, 65, 80, 95, 120]   # Price in lakhs
}

# Create DataFrame
df = pd.DataFrame(data)

# Independent variables (X)
X = df[['Area', 'Bedrooms', 'Age']]

# Dependent variable (Y)
y = df['Price']

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict price of a new house
# Example: Area=1400 sq ft, Bedrooms=3, Age=7 years
prediction = model.predict([[1400, 3, 7]])

print("Predicted House Price (in lakhs):", prediction[0])
```
